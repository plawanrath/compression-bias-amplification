# Exploration Notebook

Ad-hoc analysis for the compression bias amplification experiment.

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

from src.metrics import compute_metrics
from src.stats import chi_squared_test
```

```python
# Load raw results
raw_dir = Path("../results/raw")
rows = []
for f in raw_dir.glob("*.jsonl"):
    rows.extend(json.loads(line) for line in open(f))

df = pd.DataFrame(rows)
print(f"Loaded {len(df)} records")
df.head()
```

```python
# Compute metrics
metrics = compute_metrics(df)
metrics
```

```python
# Quick SRS plot
fig, ax = plt.subplots(figsize=(10, 6))
sns.lineplot(data=metrics, x="quant", y="srs", hue="model", style="category", markers=True, ax=ax)
ax.invert_xaxis()
ax.set_xlabel("Quantization Bit-width")
ax.set_ylabel("SRS")
ax.axhline(y=1/3, color="gray", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
